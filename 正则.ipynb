\d 数字  \d+连续数字 \D非数字 \s空白字符 \S非空白字符 \w字母，下划线，数字  \W非字母数字下划线
.任意字符

*零个或多个    +1个或者多个    ？0或1    {n} n个    {n,}至少n个   {n,m}n到m个

^：字符串开头     $：字符串结尾   ^...$：整段完全匹配   

r'\d+'
只要文本里有一串数字就行。
有 ^ 和 $：
r'^\d+$'
要求整个字符串必须全是数字。

\b单词边界
r'\bcat\b'
表示匹配完整单词 cat。


|或者  
[]匹配其中任意一个字符   
[^]不在括号里的任意字符 用于：$1,200会匹配：,

()分组：  r'(cat|dog)s?'cat 或 dog，后面可以有 s，也可以没有

反斜杠\
如果你想匹配真正的小数点，要写：r'\.'
\d+\.\d+可以匹配 3.14
