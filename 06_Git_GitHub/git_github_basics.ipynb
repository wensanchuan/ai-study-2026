## 1. Git 是什么？

Git 是一个版本控制系统。

它的作用不是简单“备份文件夹”，而是记录一个项目从开始到现在的所有关键版本。

比如一个项目可能经历：

A：创建 README  
B：添加 QC 脚本  
C：添加 UMAP 脚本  
D：新建分支尝试 DotPlot  
E：把 DotPlot 合并回主分支  

Git 会把这些版本组织成一张由 commit 组成的历史图。

## 2. Git 的三层模型

Git 日常使用时，最重要的是三层：

1. working directory：工作区
2. staging area / index：暂存区
3. repository：版本库 / commit 历史

关系是：

working directory  
→ git add  
→ staging area  
→ git commit  
→ repository

## 3. 基础命令

# 初始化仓库
git init

# 查看当前状态
git status

# 把文件加入暂存区
git add Readme.md

# 把所有修改加入暂存区
git add .

# 提交
git commit -m "first commit"

# 查看历史
git log

# 推荐查看历史图
git log --all --graph --decorate --oneline

## 9. merge 合并

```markdown
## 8. merge 合并

merge 的意思是：

把另一个分支的修改合并进当前分支。

如果我现在在 master：

```bash
git switch master
把 add-dotplot 分支的内容合并到 master。

```markdown
### 8.1 fast-forward merge

如果 master 没有产生新提交，只是 add-dotplot 比 master 多走了一步：

```text
A --- B --- C --- D
            ↑     ↑
          master  add-dotplot

执行：

git merge add-dotplot

之后：

A --- B --- C --- D
                  ↑
                master
                  ↑
             add-dotplot
                  ↑
                HEAD

这种叫 fast-forward。

本质是 master 指针直接向前移动到 D。


```markdown
### 8.2 普通 merge

如果两个分支都各自有新提交：

```text
A --- B --- C --- E    master
             \
              D        add-dotplot

执行：

git merge add-dotplot

可能生成 merge commit：

A --- B --- C --- E --- M    master
             \       /
              D ----      add-dotplot

M 有两个 parent。

# 查看远程仓库
git remote -v

# 添加远程仓库
git remote add origin git@github.com:username/repo.git

# 推送到远程
git push -u origin master

# 后续推送
git push


```markdown
### fetch

fetch 是从远程下载新的 commit 和分支信息，但不自动合并。

```bash
git fetch origin


```markdown
### pull

pull 约等于：

```bash
git fetch
git merge

只是某个文件改坏了，还没 commit：
用 git restore <file>

只是 git add 错了：
用 git restore --staged <file>

刚 commit 完，想撤销 commit 但保留修改：
用 git reset --soft HEAD~1

刚 commit 完，想撤销 commit，修改回到工作区：
用 git reset HEAD~1

想彻底回到上一个版本，丢掉当前修改：
谨慎用 git reset --hard HEAD~1

reset 错了，想找回来：
用 git reflog

git log --all --graph --decorate --oneline
