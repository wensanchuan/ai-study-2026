# awk按列处理文本的小语言
```text
常见用法 1.awk'{print $1}' 打印每一行的每一列
    awk'{print $2}'打印每一行的第二列
    $1 第一列 $2 第二列 $0整行
        2.处理CSV：指定分隔符 -F
    awk默认空格分隔，所以处理csv时
    awk -F ',' '{print $1}' metadata.csv
        3.去掉表头
    awk -F ',' 'NR>1 {print $3} xxx
    等价于
    cut -d ',' -f 3 xxx |tail -n +2 
        4.按条件筛选
    awk -F ',' '$3='diabetes'{print $0}' xxx  只看第三列（group列）是diabetes的行
        5.打印多列
    awk -F ',' '{print $1,$2,$3}' xxx
```

；  &&   ||   &   nohup   jobs   ps   kill   chmod +x   bash script.sh

```text
  ;:命令1 ; 命令2  ;:不管命令1执行成功还是失败，都继续执行命令2
  &&: 命令1 && 命令2 如果命令1成功，就执行命令2；如果命令1失败，就不执行命令2。
  ||:前一条失败，才执行后一条
  命令1 && 命令2 || 命令3 命令2在命令1成功时才执行，否则，执行命令3

  &: 把命令放在后台
    sleep 60 & 它会在后台运行 60 秒，Shell 会立刻把命令行还给你
    可能显示：[1] 12345这里：[1]     job 编号 
                            12345   进程 PID
    Rscript step01_qc.R > step01_qc.log 2>&1 &

jobs:查看当前shell的后台任务 jobs 只看当前这个 Shell 会话启动的后台任务。
fg：把后台任务拉回前台
    如果jobs显示 [1]+  Running    sleep 60 & 
    fg %1可以拉回进程1

ctrl z和ctrl c
    ctrl c：中断当前前台命令
    ctrl z：暂停当前前台命令，把它挂起 
        bg %1可以让进程1在后台继续运行

nohup：断开ssh后仍尽量继续运行 （no hang up）

tmux: 更适合长任务管理
    如果要离开tmux：ctrl + B 然后按D
    重新进入：tmux attach -t xx

ps：jobs只能看当前shell的后台任务
    ps aux | grep Rscript  列出系统进程；筛选包含 Rscript 的行。

kill：终止进程
    kill pid 
    kill -9 pid 强制执行

chmod：权限
    ls -lh xxx 可以查到xxx的信息，包括权限
    chmod +x run.sh 给执行权限

bash script.sh vs ./script.sh
    运行脚本的两种方式bash ./
    bash：不要求有执行权限，能读就能执行
    ./：要有执行权限，第一行最好有shebang（eg：#!/bin/bash）
        #!/bin/bash 是什么，即shebang；这个脚本请用 /bin/bash 来解释执行。
        很多严谨脚本会写 set -euo pipefail
            set -e 只要某条命令失败，脚本立刻停止
            set -u 使用微定义变量时，直接报错
            set -o pipefail 管道中任何一步失败，都失败
``` 


## 第九部分 变量、环境变量与$PATH
    VAR=value
    echo $VAR
    export VAR=value
    $HOME
    $PATH
    which
    type
    alias
    source
    conda activate

    1. Shell 里的变量是什么？
    2. 为什么变量前面要加 $？
    3. $HOME、$PATH 是什么？
    4. Shell 怎么知道 Rscript、python、conda 在哪里？
    5. 为什么激活 conda 前要 source？

#
    1.什么是变量?
        给变量赋值时，=两边没有空格
    2.使用变量时，加$,不然只代表变量名本身
        比如NAME=heshi
        可以使用${NAME},推荐写法，一般建议加引号 "$NAME"，如果变量内容里有空格，不加引号会被 Shell 拆开
    3.单引号与双引号的区别
        echo "hello $NAME"：hello heshi
        echo 'hello $NAME'：hello $NAME

        cat > script.R <<'EOF'
        ...
        EOF
        'EOF' 加单引号，是为了防止里面的 $ 被 Shell 提前解释。
    
    4.常见内置变量
        $HOME cd $HOME 等价于 cd ~ 
        更常见 cd "$HOME"

        $PWD 当前目录

        $USER 当前用户名

        $PATH shell找命令的搜索路径

        which 查看命令实际来自哪里
        which python 可能输出/home/heshi/conda_envs/scrna/bin/python

        type 比which更全面

        expert 把变量变成环境变量
            例如：MYVAR=hello
                  echo $MYVAR
                  但如果你启动一个子程序，它不一定能看到。如果你想让子程序也能看到这个变量，要：
                  export MYVAR=hello
            普通变量：当前 Shell 自己知道；环境变量：当前 Shell 启动的子程序也知道
